```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
df  = pd.read_csv("data.csv")
```

```python
df.info()
```

```python
df.head()
```

```python
# ------------------ CHECK FOR DUPLICATE VALUES ------------------

df.duplicated().value_counts()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
# ------------------ HANDLE MISSING VALUES ------------------
print("\nMissing Values:\n", df.isnull().sum())
```

```python
# ------------------ ENCODING ------------------
# Using manual mapping (best use when consistency needed)

gender_map = {'Male': 0, 'Female': 1, 'Other': 2}
df['gender'] = df['gender'].map(gender_map)
```

```python
df.head()
```

```python


smoking_map = {
    'never': 0,
    'former': 1,
    'not current': 2,
    'current': 3,
    'ever': 2,
    'No Info': -1
}
df['smoking_history'] = df['smoking_history'].map(smoking_map)
```

```python
# ------------------ VERIFY DATA ------------------
print("\nAfter Cleaning:")
print(df.head())
print(df.info())
```

```python
# ------------------ SAVE CLEAN DATA ------------------
df.to_csv("clean_data.csv", index=False)

print("\n✅ Clean data saved as clean_data.csv")
```
